<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
using System;
using System.Collections.Generic;
using System.Linq;

// Делегаты
public delegate void AnimalSoundHandler(string sound);
public delegate void AnimalActionHandler(Animal animal, string action);
public delegate bool AnimalFilter(Animal animal);

// Базовый класс Animal с событиями
public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public abstract string AnimalType { get; }
    
    // События
    public event AnimalSoundHandler OnMakeSound;
    public event AnimalActionHandler OnActionPerformed;
    
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }
    
    public abstract void MakeSound();
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Тип: {AnimalType}, Имя: {Name}, Возраст: {Age} лет");
    }
    
    // Метод для вызова событий
    protected virtual void RaiseOnMakeSound(string sound)
    {
        OnMakeSound?.Invoke(sound);
    }
    
    protected virtual void RaiseOnActionPerformed(string action)
    {
        OnActionPerformed?.Invoke(this, action);
    }
    
    public void PerformAction(string action)
    {
        Console.WriteLine($"{Name} выполняет действие: {action}");
        RaiseOnActionPerformed(action);
    }
}

// Производные классы
public class Dog : Animal
{
    public override string AnimalType => "Собака";
    public string Breed { get; set; }
    
    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }
    
    public override void MakeSound()
    {
        string sound = "Гав-гав!";
        Console.WriteLine($"{Name} гавкает: {sound}");
        RaiseOnMakeSound(sound);
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
    
    public void Fetch()
    {
        PerformAction("принес палку");
    }
}

public class Cat : Animal
{
    public override string AnimalType => "Кошка";
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, bool isIndoor) : base(name, age)
    {
        IsIndoor = isIndoor;
    }
    
    public override void MakeSound()
    {
        string sound = "Мяу-мяу!";
        Console.WriteLine($"{Name} мяукает: {sound}");
        RaiseOnMakeSound(sound);
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя кошка: {(IsIndoor ? "Да" : "Нет")}");
    }
    
    public void ClimbTree()
    {
        PerformAction("залез на дерево");
    }
}

public class Bird : Animal
{
    public override string AnimalType => "Птица";
    public double Wingspan { get; set; }
    
    public Bird(string name, int age, double wingspan) : base(name, age)
    {
        Wingspan = wingspan;
    }
    
    public override void MakeSound()
    {
        string sound = "Чик-чирик!";
        Console.WriteLine($"{Name} чирикает: {sound}");
        RaiseOnMakeSound(sound);
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }
    
    public void Fly()
    {
        PerformAction("полетел");
    }
}

public class Fish : Animal
{
    public override string AnimalType => "Рыба";
    public string WaterType { get; set; }
    
    public Fish(string name, int age, string waterType) : base(name, age)
    {
        WaterType = waterType;
    }
    
    public override void MakeSound()
    {
        string sound = "Буль-буль!";
        Console.WriteLine($"{Name} пускает пузыри: {sound}");
        RaiseOnMakeSound(sound);
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }
    
    public void Swim()
    {
        PerformAction("поплыл");
    }
}

// Класс для управления животными
public class AnimalManager
{
    public event Action<string> OnAnimalEvent;
    
    public void HandleAnimalSound(string sound)
    {
        Console.WriteLine($"[Менеджер] Зарегистрирован звук: {sound}");
        OnAnimalEvent?.Invoke($"Звук животного: {sound}");
    }
    
    public void HandleAnimalAction(Animal animal, string action)
    {
        Console.WriteLine($"[Менеджер] {animal.Name} выполнил: {action}");
        OnAnimalEvent?.Invoke($"Действие: {animal.Name} - {action}");
    }
}

class Program
{
    static void Main()
    {
        var manager = new AnimalManager();
        
        // Использование List<T> с лямбда-выражениями
        Console.WriteLine("=== РАБОТА С LIST<T> И ЛЯМБДА-ВЫРАЖЕНИЯМИ ===");
        
        List<Animal> animals = new List<Animal>
        {
            new Dog("Бобик", 3, "Овчарка"),
            new Cat("Мурка", 2, true),
            new Bird("Кеша", 1, 15.5),
            new Fish("Немо", 1, "Морская"),
            new Dog("Шарик", 5, "Дворняжка")
        };
        
        // Подписка на события для всех животных
        foreach (var animal in animals)
        {
            animal.OnMakeSound += manager.HandleAnimalSound;
            animal.OnActionPerformed += manager.HandleAnimalAction;
        }
        
        manager.OnAnimalEvent += (message) => 
            Console.WriteLine($"[СОБЫТИЕ] {DateTime.Now:HH:mm:ss}: {message}");
        
        // Демонстрация делегатов с лямбда-выражениями
        Console.WriteLine("\n=== ДЕЛЕГАТЫ И ФИЛЬТРАЦИЯ ===");
        
        // Делегат для фильтрации
        AnimalFilter filter = animal => animal.Age > 2;
        
        // Лямбда-выражения для различных фильтров
        var adultAnimals = animals.Where(a => a.Age > 2).ToList();
        var dogsOnly = animals.Where(a => a is Dog).ToList();
        var youngAnimals = animals.FindAll(a => a.Age <= 2);
        
        Console.WriteLine("\nВзрослые животные (старше 2 лет):");
        adultAnimals.ForEach(a => a.DisplayInfo());
        
        // Действия с использованием делегатов
        Console.WriteLine("\n=== ДЕЙСТВИЯ ЖИВОТНЫХ ===");
        Action<Animal> makeSoundAction = animal => animal.MakeSound();
        Action<Animal> displayAction = animal => animal.DisplayInfo();
        
        animals.ForEach(makeSoundAction);
        Console.WriteLine();
        animals.ForEach(displayAction);
        
        // Использование Dictionary<TKey, TValue> с событиями
        Console.WriteLine("\n=== РАБОТА С DICTIONARY<TKEY, TVALUE> ===");
        
        Dictionary<string, Animal> animalDictionary = new Dictionary<string, Animal>
        {
            {"dog1", new Dog("Рекс", 4, "Лабрадор")},
            {"cat1", new Cat("Васька", 3, false)},
            {"bird1", new Bird("Гоша", 2, 12.0)},
            {"fish1", new Fish("Золотик", 1, "Пресная")}
        };
        
        // Подписка на события для словарных животных
        foreach (var kvp in animalDictionary)
        {
            kvp.Value.OnMakeSound += manager.HandleAnimalSound;
            kvp.Value.OnActionPerformed += manager.HandleAnimalAction;
        }
        
        // Действия со словарными животными
        animalDictionary["dog1"].MakeSound();
        (animalDictionary["dog1"] as Dog)?.Fetch();
        
        // Функциональное программирование с животными
        Console.WriteLine("\n=== ФУНКЦИОНАЛЬНОЕ ПРОГРАММИРОВАНИЕ ===");
        
        // Цепочка методов с LINQ и лямбдами
        var result = animals
            .Where(a => a.Age > 1)
            .OrderBy(a => a.Name)
            .Select(a => new { a.Name, a.AnimalType, a.Age })
            .ToList();
        
        Console.WriteLine("Отсортированные животные старше 1 года:");
        result.ForEach(x => Console.WriteLine($"  {x.Name} ({x.AnimalType}, {x.Age} лет)"));
        
        // Группировка с лямбда-выражениями
        Console.WriteLine("\nГруппировка по типу животного:");
        var grouped = animals.GroupBy(a => a.AnimalType);
        grouped.ToList().ForEach(group =>
        {
            Console.WriteLine($"  {group.Key}: {group.Count()} животных");
            group.ToList().ForEach(animal => 
                Console.WriteLine($"    - {animal.Name}"));
        });
        
        // Агрегатные функции с лямбдами
        double averageAge = animals.Average(a => a.Age);
        int totalAge = animals.Sum(a => a.Age);
        Animal oldest = animals.OrderByDescending(a => a.Age).First();
        
        Console.WriteLine($"\nСтатистика:");
        Console.WriteLine($"Средний возраст: {averageAge:F1} лет");
        Console.WriteLine($"Общий возраст всех животных: {totalAge} лет");
        Console.WriteLine($"Самое старое животное: {oldest.Name} ({oldest.Age} лет)");
        
        // Сложные фильтры с составными лямбдами
        Console.WriteLine("\n=== СЛОЖНЫЕ ФИЛЬТРЫ ===");
        
        Func<Animal, bool> complexFilter = a => 
            (a.Age > 1 && a.Age < 5) || 
            a.Name.StartsWith("Б") || 
            a is Bird;
        
        var filteredAnimals = animals.Where(complexFilter).ToList();
        Console.WriteLine("Животные по сложному фильтру:");
        filteredAnimals.ForEach(a => a.DisplayInfo());
        
        // Демонстрация событий с разными животными
        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ СОБЫТИЙ ===");
        
        (animalDictionary["cat1"] as Cat)?.ClimbTree();
        (animalDictionary["bird1"] as Bird)?.Fly();
        (animalDictionary["fish1"] as Fish)?.Swim();
        
        // Отписка от событий (пример)
        animalDictionary["dog1"].OnMakeSound -= manager.HandleAnimalSound;
        Console.WriteLine("\nПосле отписки от события:");
        animalDictionary["dog1"].MakeSound();
    }
}
````
