<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

```csharp
using System;
using System.Collections.Generic;

// Определение базового класса Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive(); // Вызов метода Drive() для каждого транспортного средства
        }
    }
}


VehicleCollection<Car> cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.Add(new Motorcycle()); // Это возможно благодаря ограничению "where T : Vehicle"

cars.DisplayVehicles(); // Выводит: "Машина едет." и "Мотоцикл едет."


```

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

````
using System;
using System.Collections.Generic;

// Generic класс для вольера/террариума
public class Enclosure<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    public string Name { get; set; }
    public double Area { get; set; }
    
    public int AnimalCount => _animals.Count;
    
    public void AddAnimal(T animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"{animal.Name} добавлен в вольер '{Name}'");
    }
    
    public bool RemoveAnimal(T animal)
    {
        bool removed = _animals.Remove(animal);
        if (removed)
        {
            Console.WriteLine($"{animal.Name} удален из вольера '{Name}'");
        }
        return removed;
    }
    
    public void DisplayAnimals()
    {
        Console.WriteLine($"\nЖивотные в вольере '{Name}':");
        foreach (var animal in _animals)
        {
            animal.DisplayInfo();
        }
    }
    
    public T FindAnimal(string name)
    {
        return _animals.Find(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
}

// Generic класс для кормления
public class Feeder<T> where T : Animal
{
    public string FoodType { get; set; }
    
    public void FeedAll(Enclosure<T> enclosure)
    {
        Console.WriteLine($"\nКормление животных в вольере '{enclosure.Name}' едой: {FoodType}");
        // Здесь мог бы быть код для массового кормления
    }
    
    public void FeedAnimal(T animal)
    {
        Console.WriteLine($"Кормление {animal.Name} едой: {FoodType}");
        animal.Eat(FoodType);
    }
}

// Generic класс для медицинских процедур
public class VeterinaryClinic<T> where T : Animal
{
    private Dictionary<T, List<string>> _medicalRecords = new Dictionary<T, List<string>>();
    
    public void PerformCheckup(T animal)
    {
        Console.WriteLine($"\nМедосмотр для {animal.Name}:");
        animal.DisplayInfo();
        
        if (!_medicalRecords.ContainsKey(animal))
        {
            _medicalRecords[animal] = new List<string>();
        }
        
        string record = $"Медосмотр проведен {DateTime.Now:dd.MM.yyyy}";
        _medicalRecords[animal].Add(record);
        Console.WriteLine(record);
    }
    
    public void AddMedicalRecord(T animal, string procedure)
    {
        if (!_medicalRecords.ContainsKey(animal))
        {
            _medicalRecords[animal] = new List<string>();
        }
        
        _medicalRecords[animal].Add($"{DateTime.Now:dd.MM.yyyy}: {procedure}");
        Console.WriteLine($"Медицинская запись добавлена для {animal.Name}: {procedure}");
    }
    
    public void DisplayMedicalHistory(T animal)
    {
        if (_medicalRecords.ContainsKey(animal) && _medicalRecords[animal].Count > 0)
        {
            Console.WriteLine($"\nМедицинская история {animal.Name}:");
            foreach (var record in _medicalRecords[animal])
            {
                Console.WriteLine($"  - {record}");
            }
        }
        else
        {
            Console.WriteLine($"Для {animal.Name} нет медицинских записей");
        }
    }
}

// Generic класс для системы учета
public class AnimalRegistry<T> where T : Animal
{
    private List<T> _registeredAnimals = new List<T>();
    
    public void RegisterAnimal(T animal)
    {
        if (!_registeredAnimals.Contains(animal))
        {
            _registeredAnimals.Add(animal);
            Console.WriteLine($"Животное {animal.Name} зарегистрировано в системе");
        }
    }
    
    public void UnregisterAnimal(T animal)
    {
        if (_registeredAnimals.Remove(animal))
        {
            Console.WriteLine($"Животное {animal.Name} удалено из системы");
        }
    }
    
    public List<T> FindBySpecies(string species)
    {
        return _registeredAnimals.FindAll(a => a.Species.Equals(species, StringComparison.OrdinalIgnoreCase));
    }
    
    public void DisplayAllAnimals()
    {
        Console.WriteLine($"\nВсе зарегистрированные животные ({_registeredAnimals.Count}):");
        foreach (var animal in _registeredAnimals)
        {
            animal.DisplayInfo();
        }
    }
}

// Базовый класс Animal
public abstract class Animal
{
    private string _name;
    private int _age;
    private double _weight;
    
    // Свойства с геттерами и сеттерами
    public string Name
    {
        get => _name;
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get => _age;
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get => _weight;
        set
        {
            if (value > 0)
                _weight = Math.Round(value, 2);
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    public string Species { get; protected set; }
    public bool IsHungry { get; protected set; } = true;
    public string ID { get; private set; }
    
    protected Animal(string name, int age, double weight, string species)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Species = species;
        ID = GenerateID();
    }
    
    private string GenerateID()
    {
        return $"{Species.Substring(0, 3).ToUpper()}_{Guid.NewGuid().ToString().Substring(0, 8).ToUpper()}";
    }
    
    // Абстрактные методы
    public abstract void MakeSound();
    public abstract void Move();
    
    // Виртуальные методы
    public virtual void Eat()
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден.");
        }
    }
    
    public virtual void Eat(string food)
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест {food}.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден, но может попробовать {food}.");
        }
    }
    
    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"ID: {ID}, Имя: {Name}, Возраст: {Age}, Вес: {Weight}кг, Вид: {Species}");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained = false) 
        : base(name, age, weight, "Собака")
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом.");
    }
    
    public override void Eat(string food)
    {
        if (food.ToLower().Contains("шоколад"))
        {
            Console.WriteLine($"{Name} не может есть шоколад! Это опасно!");
            return;
        }
        base.Eat(food);
    }
    
    public void Fetch(string item = "палку")
    {
        Console.WriteLine($"{Name} приносит {item}.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Порода: {Breed}, Дрессирована: {IsTrained}");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int Lives { get; private set; } = 9;
    
    public Cat(string name, int age, double weight, bool isIndoor = true) 
        : base(name, age, weight, "Кошка")
    {
        IsIndoor = isIndoor;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        string action = IsIndoor ? "грациозно ходит" : "осторожно крадется";
        Console.WriteLine($"{Name} {action}.");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
    
    public void UseLife()
    {
        if (Lives > 0)
        {
            Lives--;
            Console.WriteLine($"{Name} использовал жизнь. Осталось: {Lives}");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Домашняя: {IsIndoor}, Жизней: {Lives}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly = true) 
        : base(name, age, weight, "Птица")
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает.");
        }
        else
        {
            Console.WriteLine($"{Name} прыгает.");
        }
    }
    
    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает с размахом крыльев {Wingspan}см.");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Размах крыльев: {Wingspan}см, Умеет летать: {CanFly}");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double SwimSpeed { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double swimSpeed) 
        : base(name, age, weight, "Рыба")
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде.");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плывет со скоростью {SwimSpeed}м/с.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип воды: {WaterType}, Скорость: {SwimSpeed}м/с");
    }
}

// Демонстрация использования
class Program
{
    static void Main()
    {
        Console.WriteLine("=== Демонстрация Generic классов с животными ===\n");
        
        // Создаем животных
        var dog1 = new Dog("Бобик", 3, 12.5, "Овчарка", true);
        var dog2 = new Dog("Шарик", 5, 8.0, "Дворняжка", false);
        var cat1 = new Cat("Мурка", 2, 4.2, true);
        var cat2 = new Cat("Васька", 4, 5.1, false);
        var bird1 = new Bird("Кеша", 1, 0.3, 25.0, true);
        var bird2 = new Bird("Пингви", 2, 15.0, 60.0, false);
        var fish1 = new Fish("Немо", 1, 0.5, "соленая", 2.5);
        
        // 1. Generic вольеры для разных типов животных
        Console.WriteLine("1. ВОЛЬЕРЫ ДЛЯ ЖИВОТНЫХ:");
        
        var dogEnclosure = new Enclosure<Dog> { Name = "Собачий вольер", Area = 50.0 };
        var catEnclosure = new Enclosure<Cat> { Name = "Кошачий дом", Area = 20.0 };
        var birdEnclosure = new Enclosure<Bird> { Name = "Птичий двор", Area = 15.0 };
        
        dogEnclosure.AddAnimal(dog1);
        dogEnclosure.AddAnimal(dog2);
        catEnclosure.AddAnimal(cat1);
        catEnclosure.AddAnimal(cat2);
        birdEnclosure.AddAnimal(bird1);
        birdEnclosure.AddAnimal(bird2);
        
        dogEnclosure.DisplayAnimals();
        catEnclosure.DisplayAnimals();
        birdEnclosure.DisplayAnimals();
        
        // 2. Generic система кормления
        Console.WriteLine("\n2. СИСТЕМА КОРМЛЕНИЯ:");
        
        var dogFeeder = new Feeder<Dog> { FoodType = "мясо и кости" };
        var catFeeder = new Feeder<Cat> { FoodType = "рыба и корм" };
        var birdFeeder = new Feeder<Bird> { FoodType = "зерно и семена" };
        
        dogFeeder.FeedAnimal(dog1);
        catFeeder.FeedAnimal(cat1);
        birdFeeder.FeedAnimal(bird1);
        
        // 3. Generic ветеринарная клиника
        Console.WriteLine("\n3. ВЕТЕРИНАРНАЯ КЛИНИКА:");
        
        var dogClinic = new VeterinaryClinic<Dog>();
        var catClinic = new VeterinaryClinic<Cat>();
        
        dogClinic.PerformCheckup(dog1);
        dogClinic.AddMedicalRecord(dog1, "Вакцинация от бешенства");
        dogClinic.AddMedicalRecord(dog1, "Обработка от паразитов");
        
        catClinic.PerformCheckup(cat1);
        catClinic.AddMedicalRecord(cat1, "Стерилизация");
        
        dogClinic.DisplayMedicalHistory(dog1);
        catClinic.DisplayMedicalHistory(cat1);
        
        // 4. Generic система регистрации
        Console.WriteLine("\n4. СИСТЕМА РЕГИСТРАЦИИ ЖИВОТНЫХ:");
        
        var animalRegistry = new AnimalRegistry<Animal>();
        animalRegistry.RegisterAnimal(dog1);
        animalRegistry.RegisterAnimal(dog2);
        animalRegistry.RegisterAnimal(cat1);
        animalRegistry.RegisterAnimal(cat2);
        animalRegistry.RegisterAnimal(bird1);
        animalRegistry.RegisterAnimal(bird2);
        animalRegistry.RegisterAnimal(fish1);
        
        animalRegistry.DisplayAllAnimals();
        
        // 5. Поиск животных по виду
        Console.WriteLine("\n5. ПОИСК ЖИВОТНЫХ ПО ВИДУ:");
        
        // Для демонстрации создадим отдельные регистры по видам
        var dogRegistry = new AnimalRegistry<Dog>();
        var catRegistry = new AnimalRegistry<Cat>();
        
        dogRegistry.RegisterAnimal(dog1);
        dogRegistry.RegisterAnimal(dog2);
        catRegistry.RegisterAnimal(cat1);
        catRegistry.RegisterAnimal(cat2);
        
        Console.WriteLine("Все собаки:");
        dogRegistry.DisplayAllAnimals();
        
        Console.WriteLine("Все кошки:");
        catRegistry.DisplayAllAnimals();
        
        // 6. Демонстрация полиморфизма с Generic
        Console.WriteLine("\n6. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА:");
        
        List<Animal> allAnimals = new List<Animal> { dog1, dog2, cat1, cat2, bird1, bird2, fish1 };
        
        foreach (var animal in allAnimals)
        {
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
        
        // 7. Работа с вольером - поиск животного
        Console.WriteLine("\n7. ПОИСК ЖИВОТНОГО В ВОЛЬЕРЕ:");
        
        var foundDog = dogEnclosure.FindAnimal("Бобик");
        if (foundDog != null)
        {
            Console.WriteLine("Найдена собака:");
            foundDog.DisplayInfo();
            foundDog.Fetch("мяч");
        }
        
        // 8. Удаление животного из вольера
        Console.WriteLine("\n8. УДАЛЕНИЕ ЖИВОТНОГО ИЗ ВОЛЬЕРА:");
        dogEnclosure.RemoveAnimal(dog2);
        dogEnclosure.DisplayAnimals();
    }
}
````

````
using System;
using System.Collections.Generic;

// Generic класс для вольера/террариума
public class Enclosure<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    public string Name { get; set; }
    public double Area { get; set; }
    
    public int AnimalCount => _animals.Count;
    
    public void AddAnimal(T animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"{animal.Name} добавлен в вольер '{Name}'");
    }
    
    public bool RemoveAnimal(T animal)
    {
        bool removed = _animals.Remove(animal);
        if (removed)
        {
            Console.WriteLine($"{animal.Name} удален из вольера '{Name}'");
        }
        return removed;
    }
    
    public void DisplayAnimals()
    {
        Console.WriteLine($"\nЖивотные в вольере '{Name}':");
        foreach (var animal in _animals)
        {
            animal.DisplayInfo();
        }
    }
    
    public T FindAnimal(string name)
    {
        return _animals.Find(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
}

// Generic класс для кормления
public class Feeder<T> where T : Animal
{
    public string FoodType { get; set; }
    
    public void FeedAll(Enclosure<T> enclosure)
    {
        Console.WriteLine($"\nКормление животных в вольере '{enclosure.Name}' едой: {FoodType}");
        // Здесь мог бы быть код для массового кормления
    }
    
    public void FeedAnimal(T animal)
    {
        Console.WriteLine($"Кормление {animal.Name} едой: {FoodType}");
        animal.Eat(FoodType);
    }
}

// Generic класс для медицинских процедур
public class VeterinaryClinic<T> where T : Animal
{
    private Dictionary<T, List<string>> _medicalRecords = new Dictionary<T, List<string>>();
    
    public void PerformCheckup(T animal)
    {
        Console.WriteLine($"\nМедосмотр для {animal.Name}:");
        animal.DisplayInfo();
        
        if (!_medicalRecords.ContainsKey(animal))
        {
            _medicalRecords[animal] = new List<string>();
        }
        
        string record = $"Медосмотр проведен {DateTime.Now:dd.MM.yyyy}";
        _medicalRecords[animal].Add(record);
        Console.WriteLine(record);
    }
    
    public void AddMedicalRecord(T animal, string procedure)
    {
        if (!_medicalRecords.ContainsKey(animal))
        {
            _medicalRecords[animal] = new List<string>();
        }
        
        _medicalRecords[animal].Add($"{DateTime.Now:dd.MM.yyyy}: {procedure}");
        Console.WriteLine($"Медицинская запись добавлена для {animal.Name}: {procedure}");
    }
    
    public void DisplayMedicalHistory(T animal)
    {
        if (_medicalRecords.ContainsKey(animal) && _medicalRecords[animal].Count > 0)
        {
            Console.WriteLine($"\nМедицинская история {animal.Name}:");
            foreach (var record in _medicalRecords[animal])
            {
                Console.WriteLine($"  - {record}");
            }
        }
        else
        {
            Console.WriteLine($"Для {animal.Name} нет медицинских записей");
        }
    }
}

// Generic класс для системы учета
public class AnimalRegistry<T> where T : Animal
{
    private List<T> _registeredAnimals = new List<T>();
    
    public void RegisterAnimal(T animal)
    {
        if (!_registeredAnimals.Contains(animal))
        {
            _registeredAnimals.Add(animal);
            Console.WriteLine($"Животное {animal.Name} зарегистрировано в системе");
        }
    }
    
    public void UnregisterAnimal(T animal)
    {
        if (_registeredAnimals.Remove(animal))
        {
            Console.WriteLine($"Животное {animal.Name} удалено из системы");
        }
    }
    
    public List<T> FindBySpecies(string species)
    {
        return _registeredAnimals.FindAll(a => a.Species.Equals(species, StringComparison.OrdinalIgnoreCase));
    }
    
    public void DisplayAllAnimals()
    {
        Console.WriteLine($"\nВсе зарегистрированные животные ({_registeredAnimals.Count}):");
        foreach (var animal in _registeredAnimals)
        {
            animal.DisplayInfo();
        }
    }
}

// Базовый класс Animal
public abstract class Animal
{
    private string _name;
    private int _age;
    private double _weight;
    
    // Свойства с геттерами и сеттерами
    public string Name
    {
        get => _name;
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get => _age;
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get => _weight;
        set
        {
            if (value > 0)
                _weight = Math.Round(value, 2);
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    public string Species { get; protected set; }
    public bool IsHungry { get; protected set; } = true;
    public string ID { get; private set; }
    
    protected Animal(string name, int age, double weight, string species)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Species = species;
        ID = GenerateID();
    }
    
    private string GenerateID()
    {
        return $"{Species.Substring(0, 3).ToUpper()}_{Guid.NewGuid().ToString().Substring(0, 8).ToUpper()}";
    }
    
    // Абстрактные методы
    public abstract void MakeSound();
    public abstract void Move();
    
    // Виртуальные методы
    public virtual void Eat()
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден.");
        }
    }
    
    public virtual void Eat(string food)
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест {food}.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден, но может попробовать {food}.");
        }
    }
    
    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"ID: {ID}, Имя: {Name}, Возраст: {Age}, Вес: {Weight}кг, Вид: {Species}");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained = false) 
        : base(name, age, weight, "Собака")
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом.");
    }
    
    public override void Eat(string food)
    {
        if (food.ToLower().Contains("шоколад"))
        {
            Console.WriteLine($"{Name} не может есть шоколад! Это опасно!");
            return;
        }
        base.Eat(food);
    }
    
    public void Fetch(string item = "палку")
    {
        Console.WriteLine($"{Name} приносит {item}.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Порода: {Breed}, Дрессирована: {IsTrained}");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int Lives { get; private set; } = 9;
    
    public Cat(string name, int age, double weight, bool isIndoor = true) 
        : base(name, age, weight, "Кошка")
    {
        IsIndoor = isIndoor;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        string action = IsIndoor ? "грациозно ходит" : "осторожно крадется";
        Console.WriteLine($"{Name} {action}.");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
    
    public void UseLife()
    {
        if (Lives > 0)
        {
            Lives--;
            Console.WriteLine($"{Name} использовал жизнь. Осталось: {Lives}");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Домашняя: {IsIndoor}, Жизней: {Lives}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly = true) 
        : base(name, age, weight, "Птица")
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает.");
        }
        else
        {
            Console.WriteLine($"{Name} прыгает.");
        }
    }
    
    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает с размахом крыльев {Wingspan}см.");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Размах крыльев: {Wingspan}см, Умеет летать: {CanFly}");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double SwimSpeed { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double swimSpeed) 
        : base(name, age, weight, "Рыба")
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде.");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плывет со скоростью {SwimSpeed}м/с.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип воды: {WaterType}, Скорость: {SwimSpeed}м/с");
    }
}

// Демонстрация использования
class Program
{
    static void Main()
    {
        Console.WriteLine("=== Демонстрация Generic классов с животными ===\n");
        
        // Создаем животных
        var dog1 = new Dog("Бобик", 3, 12.5, "Овчарка", true);
        var dog2 = new Dog("Шарик", 5, 8.0, "Дворняжка", false);
        var cat1 = new Cat("Мурка", 2, 4.2, true);
        var cat2 = new Cat("Васька", 4, 5.1, false);
        var bird1 = new Bird("Кеша", 1, 0.3, 25.0, true);
        var bird2 = new Bird("Пингви", 2, 15.0, 60.0, false);
        var fish1 = new Fish("Немо", 1, 0.5, "соленая", 2.5);
        
        // 1. Generic вольеры для разных типов животных
        Console.WriteLine("1. ВОЛЬЕРЫ ДЛЯ ЖИВОТНЫХ:");
        
        var dogEnclosure = new Enclosure<Dog> { Name = "Собачий вольер", Area = 50.0 };
        var catEnclosure = new Enclosure<Cat> { Name = "Кошачий дом", Area = 20.0 };
        var birdEnclosure = new Enclosure<Bird> { Name = "Птичий двор", Area = 15.0 };
        
        dogEnclosure.AddAnimal(dog1);
        dogEnclosure.AddAnimal(dog2);
        catEnclosure.AddAnimal(cat1);
        catEnclosure.AddAnimal(cat2);
        birdEnclosure.AddAnimal(bird1);
        birdEnclosure.AddAnimal(bird2);
        
        dogEnclosure.DisplayAnimals();
        catEnclosure.DisplayAnimals();
        birdEnclosure.DisplayAnimals();
        
        // 2. Generic система кормления
        Console.WriteLine("\n2. СИСТЕМА КОРМЛЕНИЯ:");
        
        var dogFeeder = new Feeder<Dog> { FoodType = "мясо и кости" };
        var catFeeder = new Feeder<Cat> { FoodType = "рыба и корм" };
        var birdFeeder = new Feeder<Bird> { FoodType = "зерно и семена" };
        
        dogFeeder.FeedAnimal(dog1);
        catFeeder.FeedAnimal(cat1);
        birdFeeder.FeedAnimal(bird1);
        
        // 3. Generic ветеринарная клиника
        Console.WriteLine("\n3. ВЕТЕРИНАРНАЯ КЛИНИКА:");
        
        var dogClinic = new VeterinaryClinic<Dog>();
        var catClinic = new VeterinaryClinic<Cat>();
        
        dogClinic.PerformCheckup(dog1);
        dogClinic.AddMedicalRecord(dog1, "Вакцинация от бешенства");
        dogClinic.AddMedicalRecord(dog1, "Обработка от паразитов");
        
        catClinic.PerformCheckup(cat1);
        catClinic.AddMedicalRecord(cat1, "Стерилизация");
        
        dogClinic.DisplayMedicalHistory(dog1);
        catClinic.DisplayMedicalHistory(cat1);
        
        // 4. Generic система регистрации
        Console.WriteLine("\n4. СИСТЕМА РЕГИСТРАЦИИ ЖИВОТНЫХ:");
        
        var animalRegistry = new AnimalRegistry<Animal>();
        animalRegistry.RegisterAnimal(dog1);
        animalRegistry.RegisterAnimal(dog2);
        animalRegistry.RegisterAnimal(cat1);
        animalRegistry.RegisterAnimal(cat2);
        animalRegistry.RegisterAnimal(bird1);
        animalRegistry.RegisterAnimal(bird2);
        animalRegistry.RegisterAnimal(fish1);
        
        animalRegistry.DisplayAllAnimals();
        
        // 5. Поиск животных по виду
        Console.WriteLine("\n5. ПОИСК ЖИВОТНЫХ ПО ВИДУ:");
        
        // Для демонстрации создадим отдельные регистры по видам
        var dogRegistry = new AnimalRegistry<Dog>();
        var catRegistry = new AnimalRegistry<Cat>();
        
        dogRegistry.RegisterAnimal(dog1);
        dogRegistry.RegisterAnimal(dog2);
        catRegistry.RegisterAnimal(cat1);
        catRegistry.RegisterAnimal(cat2);
        
        Console.WriteLine("Все собаки:");
        dogRegistry.DisplayAllAnimals();
        
        Console.WriteLine("Все кошки:");
        catRegistry.DisplayAllAnimals();
        
        // 6. Демонстрация полиморфизма с Generic
        Console.WriteLine("\n6. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА:");
        
        List<Animal> allAnimals = new List<Animal> { dog1, dog2, cat1, cat2, bird1, bird2, fish1 };
        
        foreach (var animal in allAnimals)
        {
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
        
        // 7. Работа с вольером - поиск животного
        Console.WriteLine("\n7. ПОИСК ЖИВОТНОГО В ВОЛЬЕРЕ:");
        
        var foundDog = dogEnclosure.FindAnimal("Бобик");
        if (foundDog != null)
        {
            Console.WriteLine("Найдена собака:");
            foundDog.DisplayInfo();
            foundDog.Fetch("мяч");
        }
        
        // 8. Удаление животного из вольера
        Console.WriteLine("\n8. УДАЛЕНИЕ ЖИВОТНОГО ИЗ ВОЛЬЕРА:");
        dogEnclosure.RemoveAnimal(dog2);
        dogEnclosure.DisplayAnimals();
    }
}
````
