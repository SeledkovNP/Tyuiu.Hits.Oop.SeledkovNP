### Перекрытие методов в C#

Перекрытие методов (или переопределение) — это процесс, при котором производный класс предоставляет новую реализацию метода, уже существующего в его базовом классе. Это позволяет изменить или расширить поведение метода в производном классе без изменения исходного метода в базовом классе. Перекрытие методов является ключевой особенностью полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, где каждый из этих классов имеет метод `Drive()`, который принимает параметр `speed`.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive()
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

В этом примере, метод `Drive()` в классе `Car` и `Motorcycle` переопределяет метод `Drive()` из базового класса `Vehicle`. Это означает, что при вызове метода `Drive()` на объекте типа `Car` или `Motorcycle`, будет вызвана соответствующая реализация метода в этих классах, а не в классе `Vehicle`.

#### Отличие перекрытия от перегрузки методов

Перегрузка методов — это процесс определения нескольких методов с одним и тем же именем, но с разным количеством или типом параметров. Например:

```csharp
public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine(number);
    }

    public void Print(double number)
    {
        Console.WriteLine(number.ToString());
    }
}
```

В этом случае, у нас есть два метода с именем `Print`, но с разным типом параметра. Это отличается от перекрытия методов, где метод в производном классе полностью заменяет метод в базовом классе.

#### Вызов методов родителя

Иногда в производном классе может потребоваться вызвать метод родителя. Для этого используется ключевое слово `base`. Например, если мы хотим добавить дополнительную логику в метод `Drive()` класса `Car`, сохраняя при этом оригинальное поведение метода `Drive()` из класса `Vehicle`.

```csharp
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed); // Вызов метода Drive() из класса Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}
```

В этом примере, после вызова `base.Drive(speed)`, выполняется дополнительная логика, специфичная для класса `Car`.


### Перегрузка методов в C#

Перегрузка методов — это способность класса определить несколько методов с одним и тем же именем, но с разным количеством или типом параметров. Это позволяет разработчикам предоставлять различные варианты выполнения одного и того же действия в зависимости от входных данных. Перегрузка методов является одной из форм выражения полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример, где класс `Vehicle` содержит перегруженные методы `Drive()`, которые принимают разные типы параметров.

```csharp
public class Vehicle
{
    // Метод Drive(), принимающий int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Метод Drive(), принимающий string
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Метод Drive(), принимающий bool
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
        }
    }
}
```

В этом примере класс `Vehicle` определяет три метода с именем `Drive()`, каждый из которых принимает разные типы параметров: `int`, `string` и `bool`. Это позволяет вызывать разные версии метода `Drive()` в зависимости от контекста, например:

```csharp
Vehicle myVehicle = new Vehicle();

myVehicle.Drive(60); // Вызов метода Drive() с int параметром
myVehicle.Drive("Eco"); // Вызов метода Drive() с string параметром
myVehicle.Drive(true); // Вызов метода Drive() с bool параметром
```

#### Отличие перегрузки от переопределения

Отличие перегрузки методов от переопределения заключается в том, что перегрузка методов включает определение нескольких методов с одним именем, но с разным набором параметров, тогда как переопределение методов (или перекрытие) означает предоставление новой реализации метода в производном классе, который уже существует в базовом классе.

#### Вызов методов

При вызове перегруженного метода компилятор выбирает подходящую версию метода на основе типов и количества аргументов, переданных в вызове. Если типы и количество аргументов совпадают с несколькими версиями метода, компилятор выберет ту версию, которая была объявлена первой.

### Заключение

Перегрузка методов в C# позволяет классам предоставлять различные варианты выполнения одного и того же действия, адаптируясь к различным условиям использования. Это мощный инструмент для создания гибких и удобных API, позволяющих клиентам класса использовать его функциональность наиболее эффективно.

Перекрытие методов в C# позволяет производным классам изменять или расширять поведение методов базовых классов, что является ключевой особенностью полиморфизма. Это отличается от перегрузки методов, которая предполагает определение нескольких методов с одинаковым именем, но с разным набором параметров.

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также используйте перекрытия и перегрузки методов.

````
using System;
using System.Collections.Generic;

// Базовый класс Animal
public class Animal
{
    // Поля с свойствами (геттеры и сеттеры)
    private string _name;
    private int _age;
    private double _weight;
    
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    public string Species { get; protected set; }
    public bool IsHungry { get; protected set; } = true;

    // Конструкторы
    public Animal() : this("Неизвестно", 0, 1.0, "Неопределенный вид") { }
    
    public Animal(string name, int age, double weight, string species)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Species = species;
    }

    // Виртуальные методы для переопределения
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }
    
    public virtual void Eat()
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден.");
        }
    }
    
    // Перегруженные методы Eat
    public virtual void Eat(string food)
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест {food}.");
            IsHungry = false;
        }
        else
        {
            Console.WriteLine($"{Name} не голоден, но может попробовать {food}.");
        }
    }
    
    public virtual void Eat(string food, double amount)
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест {amount} кг {food}.");
            IsHungry = false;
            Weight += amount * 0.1; // Увеличиваем вес
        }
        else
        {
            Console.WriteLine($"{Name} не голоден, но может попробовать {amount} кг {food}.");
        }
    }
    
    public virtual void Move()
    {
        Console.WriteLine($"{Name} перемещается.");
    }
    
    // Перегруженный метод Move
    public virtual void Move(int distance)
    {
        Console.WriteLine($"{Name} перемещается на {distance} метров.");
    }
    
    // Виртуальный метод для получения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Вес: {Weight} кг, Вид: {Species}, Голоден: {IsHungry}");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog() : base()
    {
        Species = "Собака";
        Breed = "Неизвестная порода";
    }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained = false) 
        : base(name, age, weight, "Собака")
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    // Переопределение методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом.");
    }
    
    public override void Move(int distance)
    {
        string activity = distance > 100 ? "долго бегает" : "прогуливается";
        Console.WriteLine($"{Name} {activity} на {distance} метров.");
    }
    
    // Переопределение метода Eat с дополнительной логикой
    public override void Eat(string food)
    {
        if (food.ToLower().Contains("шоколад"))
        {
            Console.WriteLine($"{Name} не может есть шоколад! Это опасно для собак!");
            return;
        }
        
        base.Eat(food);
        
        if (food.ToLower().Contains("кость"))
        {
            Console.WriteLine($"{Name} очень доволен костью!");
        }
    }
    
    // Новый метод, специфичный для собак
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку.");
    }
    
    // Перегруженный метод Fetch
    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} приносит {item}.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Дрессирована: {IsTrained}");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public int Lives { get; private set; } = 9;
    public bool IsIndoor { get; set; }
    
    public Cat() : base()
    {
        Species = "Кошка";
    }
    
    public Cat(string name, int age, double weight, bool isIndoor = true) 
        : base(name, age, weight, "Кошка")
    {
        IsIndoor = isIndoor;
    }
    
    // Переопределение методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        string movement = IsIndoor ? "грациозно ходит по дому" : "осторожно крадется";
        Console.WriteLine($"{Name} {movement}.");
    }
    
    // Переопределение метода Eat
    public override void Eat(string food)
    {
        if (food.ToLower().Contains("рыба"))
        {
            Console.WriteLine($"{Name} с удовольствием ест рыбу! Мур-мур!");
            IsHungry = false;
            Weight += 0.2;
        }
        else
        {
            base.Eat(food);
        }
    }
    
    // Новый метод, специфичный для кошек
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мррррр...");
    }
    
    // Перегруженный метод Purr
    public void Purr(int volume)
    {
        string soundLevel = volume > 5 ? "громко" : "тихо";
        Console.WriteLine($"{Name} {soundLevel} мурлычет: МРРРРР...");
    }
    
    public void UseLife()
    {
        if (Lives > 1)
        {
            Lives--;
            Console.WriteLine($"{Name} использовал одну жизнь. Осталось жизней: {Lives}");
        }
        else
        {
            Console.WriteLine($"{Name} использовал последнюю жизнь!");
            Lives = 0;
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Жизней: {Lives}, Домашняя: {IsIndoor}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird() : base()
    {
        Species = "Птица";
        CanFly = true;
    }
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly = true) 
        : base(name, age, weight, "Птица")
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    // Переопределение методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает в небе.");
        }
        else
        {
            Console.WriteLine($"{Name} прыгает по земле.");
        }
    }
    
    public override void Move(int distance)
    {
        if (CanFly && distance > 50)
        {
            Console.WriteLine($"{Name} пролетает {distance} метров по воздуху.");
        }
        else
        {
            Console.WriteLine($"{Name} перемещается на {distance} метров по земле.");
        }
    }
    
    // Переопределение метода Eat
    public override void Eat(string food, double amount)
    {
        if (food.ToLower().Contains("зерно") || food.ToLower().Contains("семена"))
        {
            Console.WriteLine($"{Name} клюет {amount} кг {food}.");
            IsHungry = false;
            Weight += amount * 0.05;
        }
        else
        {
            Console.WriteLine($"{Name} не любит {food}. Птицы предпочитают зерно и семена.");
        }
    }
    
    // Новый метод, специфичный для птиц
    public void Fly()
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} взлетает и парит в воздухе с размахом крыльев {Wingspan} см.");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать.");
        }
    }
    
    // Перегруженный метод Fly
    public void Fly(int height)
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} летает на высоте {height} метров.");
        }
        else
        {
            Console.WriteLine($"{Name} не умеет летать.");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см, Умеет летать: {CanFly}");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; } // "пресная" или "соленая"
    public double SwimSpeed { get; set; }
    
    public Fish() : base()
    {
        Species = "Рыба";
        WaterType = "пресная";
    }
    
    public Fish(string name, int age, double weight, string waterType, double swimSpeed) 
        : base(name, age, weight, "Рыба")
    {
        WaterType = waterType;
        SwimSpeed = swimSpeed;
    }
    
    // Переопределение методов
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void Move()
    {
        string speedDesc = SwimSpeed > 10 ? "быстро плавает" : "медленно плавает";
        Console.WriteLine($"{Name} {speedDesc} в {WaterType} воде.");
    }
    
    public override void Move(int distance)
    {
        double time = distance / SwimSpeed;
        Console.WriteLine($"{Name} проплывет {distance} метров за {time:F1} секунд.");
    }
    
    // Переопределение метода Eat
    public override void Eat()
    {
        if (IsHungry)
        {
            Console.WriteLine($"{Name} ест планктон и водоросли.");
            IsHungry = false;
            Weight += 0.05;
        }
        else
        {
            Console.WriteLine($"{Name} не голодна.");
        }
    }
    
    // Новый метод, специфичный для рыб
    public void Swim()
    {
        Console.WriteLine($"{Name} грациозно плавает со скоростью {SwimSpeed} м/с.");
    }
    
    // Перегруженный метод Swim
    public void Swim(string direction)
    {
        Console.WriteLine($"{Name} плывет {direction} со скоростью {SwimSpeed} м/с.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}, Скорость плавания: {SwimSpeed} м/с");
    }
}

// Демонстрация использования
class Program
{
    static void Main()
    {
        Console.WriteLine("=== Демонстрация полиморфизма с животными ===\n");
        
        // Создаем список животных разных типов
        List<Animal> animals = new List<Animal>
        {
            new Dog("Бобик", 3, 12.5, "Овчарка", true),
            new Cat("Мурка", 2, 4.2, true),
            new Bird("Кеша", 1, 0.3, 25.0, true),
            new Fish("Немо", 1, 0.5, "соленая", 2.5),
            new Dog("Шарик", 5, 8.0, "Дворняжка"),
            new Bird("Пингви", 2, 15.0, 60.0, false)
        };
        
        // Демонстрация полиморфизма - общие методы
        Console.WriteLine("1. Демонстрация переопределенных методов:\n");
        foreach (var animal in animals)
        {
            animal.MakeSound(); // Каждое животное издает свой звук
        }
        
        Console.WriteLine("\n2. Демонстрация движения:\n");
        foreach (var animal in animals)
        {
            animal.Move(); // Каждое животное движется по-своему
        }
        
        Console.WriteLine("\n3. Демонстрация перегруженных методов:\n");
        // Используем перегруженные методы с разными параметрами
        animals[0].Move(50);  // Собака бегает
        animals[2].Move(100); // Птица летает
        animals[3].Move(30);  // Рыба плывет
        
        Console.WriteLine("\n4. Демонстрация питания с разными параметрами:\n");
        // Разные способы вызова метода Eat
        animals[0].Eat();                     // Без параметров
        animals[1].Eat("рыба");               // С указанием еды
        animals[2].Eat("зерно", 0.1);         // С едой и количеством
        animals[3].Eat();                     // Рыба ест по-своему
        
        Console.WriteLine("\n5. Демонстрация специфических методов:\n");
        // Приведение типов для вызова специфических методов
        foreach (var animal in animals)
        {
            if (animal is Dog dog)
            {
                dog.Fetch("мяч");
            }
            else if (animal is Cat cat)
            {
                cat.Purr(7);
            }
            else if (animal is Bird bird)
            {
                bird.Fly(20);
            }
            else if (animal is Fish fish)
            {
                fish.Swim("вперед");
            }
        }
        
        Console.WriteLine("\n6. Информация о всех животных:\n");
        foreach (var animal in animals)
        {
            animal.DisplayInfo();
            Console.WriteLine();
        }
        
        // Дополнительная демонстрация с кошкой
        Console.WriteLine("7. Демонстрация работы с кошкой:\n");
        Cat myCat = new Cat("Васька", 4, 5.0, false);
        myCat.DisplayInfo();
        myCat.MakeSound();
        myCat.Eat("корм");
        myCat.UseLife();
        myCat.Purr();
        
        // Демонстрация работы с собакой и разной едой
        Console.WriteLine("\n8. Демонстрация питания собаки разной едой:\n");
        Dog myDog = new Dog("Рекс", 2, 10.0, "Лабрадор");
        myDog.Eat("кость");        // Обычная еда
        myDog.Eat("шоколад");      // Опасная еда
        myDog.Eat("мясо", 0.5);    // Еда с количеством
    }
}
````
