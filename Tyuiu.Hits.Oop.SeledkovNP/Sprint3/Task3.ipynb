```
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый класс Supplier
public class Supplier
{
    public int Id { get; set; }
    public string Name { get; set; }
    public string Phone { get; set; }
    public string Email { get; set; }
    public decimal Rating { get; set; }
    public bool IsActive { get; set; } = true;
    public DateTime Created { get; set; } = DateTime.Now;

    public Supplier(int id, string name, string phone, string email)
    {
        Id = id;
        Name = name;
        Phone = phone;
        Email = email;
    }

    public virtual void Display() => Console.WriteLine($"ID: {Id}, {Name}, Рейтинг: {Rating}");

    public virtual decimal GetDiscount(int amount) => amount > 10000 ? 0.05m : 0m;

    public virtual void ProcessOrder(string order, decimal amount)
    {
        Console.WriteLine($"{Name} обрабатывает заказ: {order}");
        Console.WriteLine($"Скидка: {GetDiscount((int)amount):P0}");
    }
}

// Поставщик товаров
public class ProductSupplier : Supplier
{
    public string Category { get; set; }
    public int Stock { get; set; }
    public List<string> Products { get; } = new List<string>();

    public ProductSupplier(int id, string name, string phone, string email, string category) 
        : base(id, name, phone, email) 
    {
        Category = category;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Категория: {Category}, Товаров: {Products.Count}");
    }

    public override decimal GetDiscount(int amount)
    {
        decimal discount = base.GetDiscount(amount);
        if (amount > 50000) discount += 0.08m;
        return discount;
    }

    public void AddProduct(string product)
    {
        Products.Add(product);
        Console.WriteLine($"{Name} добавил товар: {product}");
    }
}

// Поставщик услуг
public class ServiceSupplier : Supplier
{
    public string ServiceType { get; set; }
    public decimal HourlyRate { get; set; }
    public List<string> Areas { get; } = new List<string>();

    public ServiceSupplier(int id, string name, string phone, string email, string serviceType) 
        : base(id, name, phone, email)
    {
        ServiceType = serviceType;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Услуга: {ServiceType}, Ставка: {HourlyRate:C}/час");
    }

    public override decimal GetDiscount(int amount) => amount > 20000 ? 0.15m : 0.1m;

    public void AddArea(string area)
    {
        Areas.Add(area);
        Console.WriteLine($"{Name} добавил зону: {area}");
    }
}

// Экологичный поставщик
public class EcoSupplier : ProductSupplier
{
    public bool IsEcoCertified { get; set; }
    public decimal CarbonFootprint { get; set; }

    public EcoSupplier(int id, string name, string phone, string email, string category) 
        : base(id, name, phone, email, category) { }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Эко-сертификат: {IsEcoCertified}, CO2: {CarbonFootprint}кг");
    }

    public override decimal GetDiscount(int amount)
    {
        decimal discount = base.GetDiscount(amount);
        if (IsEcoCertified) discount += 0.03m;
        return discount;
    }
}

// Демонстрация
class Program
{
    static void Main()
    {
        var suppliers = new List<Supplier>
        {
            new Supplier(1, "БазовыйПоставщик", "+79991112233", "base@mail.ru"),
            new ProductSupplier(2, "ТоварыОПТ", "+79994445566", "opt@mail.ru", "Электроника"),
            new ServiceSupplier(3, "УслугиПРО", "+79997778899", "pro@mail.ru", "IT"),
            new EcoSupplier(4, "ЭкоТовары", "+79993334455", "eco@mail.ru", "Органика")
        };

        // Настройка поставщиков
        ((ProductSupplier)suppliers[1]).AddProduct("Ноутбуки");
        ((ProductSupplier)suppliers[1]).AddProduct("Телефоны");
        ((ServiceSupplier)suppliers[2]).HourlyRate = 1000;
        ((ServiceSupplier)suppliers[2]).AddArea("Москва");
        ((EcoSupplier)suppliers[3]).IsEcoCertified = true;
        ((EcoSupplier)suppliers[3]).CarbonFootprint = 0.5m;

        // Демонстрация полиморфизма
        Console.WriteLine("====== ИНФОРМАЦИЯ О ПОСТАВЩИКАХ ======");
        suppliers.ForEach(s => { s.Display(); Console.WriteLine(); });

        Console.WriteLine("====== ОБРАБОТКА ЗАКАЗОВ ======");
        decimal[] amounts = { 5000, 25000, 60000, 30000 };
        for (int i = 0; i < suppliers.Count; i++)
        {
            suppliers[i].ProcessOrder($"Заказ #{i+1}", amounts[i]);
            Console.WriteLine();
        }
    }
}


/* 
====== ИНФОРМАЦИЯ О ПОСТАВЩИКАХ ======
ID: 1, БазовыйПоставщик, Рейтинг: 0

ТоварыОПТ добавил товар: Ноутбуки
ТоварыОПТ добавил товар: Телефоны
ID: 2, ТоварыОПТ, Рейтинг: 0
Категория: Электроника, Товаров: 2

ID: 3, УслугиПРО, Рейтинг: 0
Услуга: IT, Ставка: ₽1,000.00/час

УслугиПРО добавил зону: Москва
ID: 4, ЭкоТовары, Рейтинг: 0
Категория: Органика, Товаров: 0
Эко-сертификат: True, CO2: 0.5кг

====== ОБРАБОТКА ЗАКАЗОВ ======
БазовыйПоставщик обрабатывает заказ: Заказ #1
Скидка: 0%

ТоварыОПТ обрабатывает заказ: Заказ #2
Скидка: 5%

ТоварыОПТ обрабатывает заказ: Заказ #3
Скидка: 13%

УслугиПРО обрабатывает заказ: Заказ #4
Скидка: 15%

*/




```
